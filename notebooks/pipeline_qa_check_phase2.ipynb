# Pipeline QA check — phase2 (JupyterLab)

Interactive dashboard for one **phase2** observation day under
`/lustre/pipeline/exopipe/phase2/`. Runs live in `Science_*` directories (not `Run_*`).

Per run:

- **Thermal noise:** single plot at `QA/*_thermal_noise_vs_freq.png` (not a subband grid).
- **Flux ratios:** Bokeh heatmaps from `QA/*_flux_check_hybrid.csv` (one CSV per run).
- **Dewarp summary:** LST × frequency heatmap from `QA/*dewarp_summary.csv` (`freq_mhz`, `median_shift_arcmin`).
- **Stokes review:** dewarped FITS (`*I-NoTaper-*-Robust-0-*` / `*V-Taper-*-Robust-0-*`
  `image.pbcorr_dewarped.fits`) converted to QA Zarr when missing.

Phase1 Wideband products: `pipeline_qa_check_phase1.ipynb`.


```python
from pathlib import Path

from ovro_lwa_portal.viz.pipeline_qa import PipelineQAConfig

# Edit these before running the dashboard.
PIPELINE_ROOT = Path("/lustre/pipeline/exopipe/phase2/")
SYMLINK_ROOT = Path("/lustre/claw")
ZARR_ROOT = Path("/fast/claw")
I_FITS_GLOB = "*I-NoTaper-*-Robust-0-*-image.pbcorr_dewarped.fits"
V_FITS_GLOB = "*V-Taper-*-Robust-0-*-image.pbcorr_dewarped.fits"
RUN_DIR_PREFIX = "Science_"
RUN_DIR_PATTERN = r"Science_(\d{8})_(\d{6})"
QA_THERMAL_NOISE_GLOB = "QA/*_thermal_noise_vs_freq.png"
FLUX_CHECK_CSV_GLOB = "QA/*_flux_check_hybrid.csv"
# Distinct from phase1 Zarr names; none should exist until you convert FITS → Zarr.
I_QA_ZARR_STEM = "pipelineQA-phase2-I-NoTaper-Robust-0"
V_QA_ZARR_STEM = "pipelineQA-phase2-V-Taper-Robust-0"

QA_CONFIG = PipelineQAConfig(
    pipeline_root=PIPELINE_ROOT,
    symlink_root=SYMLINK_ROOT,
    zarr_root=ZARR_ROOT,
    i_fits_glob=I_FITS_GLOB,
    v_fits_glob=V_FITS_GLOB,
    run_dir_prefix=RUN_DIR_PREFIX,
    run_dir_pattern=RUN_DIR_PATTERN,
    qa_thermal_noise_glob=QA_THERMAL_NOISE_GLOB,
    flux_check_csv_glob=FLUX_CHECK_CSV_GLOB,
    flux_check_csv_per_run=True,
    i_qa_zarr_stem=I_QA_ZARR_STEM,
    v_qa_zarr_stem=V_QA_ZARR_STEM,
    thermal_noise_grid_cols=4,
    dewarp_summary_csv_glob="QA/*dewarp_summary.csv",
    thermal_noise_plot_name="thermal_noise_vs_freq",
    qa_run_label="Science",
)
```

```python
import panel as pn

from ovro_lwa_portal.viz.pipeline_qa_app import display_pipeline_qa_app

pn.extension("tabulator")
display_pipeline_qa_app(qa_config=QA_CONFIG)
```
